```python
import numpy as np
A = np.array([3, 1, 4, 2])
B = np.array([3, -1, 3, 2])
print(A)
print(B)
```

```python
A + B
```

```python
A - B
```

```python
A * B
```

```python
A / B
```

```python
np.add(A, B)  # Metoda add() dodaje tablice do siebie
```

```python
A + 3
```

```python
2 * A
```

```python
A + 3 * B
```

```python
np.subtract(A, B) # Metoda subrtact() odejmuje tablice
```

```python
np.multiply(A, B) # Metoda multiply() mnoży tablice
```

```python
np.divide(A, B) # Metoda divide dzieli tablice
```

```python
X = np.array([[1, 3], [-2, 0]])
Y = np.array([[6, 0], [-1, 2]])
print(X, '\n')
print(Y)
```

```python
X * Y
```

```python
np.dot(X, Y) # Mnożenie macieżowe dwóch macierzy
```

```python
X.dot(Y) # Mnożenie macierzowe dwóch macierzy
```

```python
Y.dot(X) # Mnozenie macierzy nie jest przemienne
```

```python
X @ Y # Inny sposób mnożenia macierzowego
```
