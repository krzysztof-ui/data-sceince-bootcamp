```python
A = np.arange(start=-10, stop=10, step=0.5)
A = A.reshape(10, -1)
A
```

```python
A > 0 # Jesli wartość większa od zera to True
```

```python
A[A > 0] # Wyświetla tylko wartości dodatnie
```

```python
np.bitwise_and(A > -5, A < 5) # Uzyskujemy maskę logiczną dla wyrażenia A > -5, A < 5
```

```python
A[np.bitwise_and(A > -5, A < 5)] # dla uzyskanej maski logicznej wyświetlamy elementy tablicy spełniające warunki
```

```python
np.bitwise_or(A < -5, A > 5 )
```

```python
A[np.bitwise_or(A < -5, A > 5 )]
```
