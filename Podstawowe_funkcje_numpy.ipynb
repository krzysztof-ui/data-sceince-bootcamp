```python
import numpy as np
```

```python
np.exp(1)
```

```python
np.sqrt(9)
```

```python
np.all([2, 3, 0]) # jeśli wszystkie elementy są True to zwraca True, jeśli chociaż jeden z nich jest False to zwraca False
```

```python
np.any([0, 0, 1]) # any() sprawdza czy jakikolwiek obiekt jest True, jeśli same zera to False, jesli chociaż jeden różny od zera to True
```

```python
bool(1.3) # pythonowa funkcja bool() zwraca False dla 0 i True dla każdej innej liczby różnej od zera.
```

```python
A = np.random.rand(5) # Wartości losowe z rozkładu jednostajnego
A
```

```python
np.argmax(A) # zwraca index pod którym występuje maksymalna wartość
```

```python
A[np.argmax(A)]
```

```python
np.argmin(A) # zwraca index pod którym występuje najmniejsza wartość
```

```python
np.argsort(A) # sortuje rosnąco podając indexy dla danych wartości
```

```python
np.max(A)  # zwraca index pod którym występuje maksymalna wartość w tablicy A
```

```python
np.min(A) # zwraca index pod którym występuje najmniejsza wartość w tablicy A
```

```python
np.mean(A) # oblicza średnią z tablicy A
```

```python
np.median(A) # oblicza medianę z tablicy A
```

```python
np.std(A) # oblicza odchylenie standardowe
```
